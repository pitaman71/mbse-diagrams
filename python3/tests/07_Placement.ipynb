# Suite PLC — Physical placement

`Placement` gives a diagram its coordinates, in diagram units, one mutation at a time, by translating the rules a mutation touches into linear constraints that an engine, Kiwi's, solves incrementally: each shape's points from its vectors' lengths and its unit length, its frame from its extent points and its grid, each grid from its children's sizes with collapsing margins, and the diagram the size of its grid. What it writes is a mutation, an upsert the store runs; each step asserts the frames and placements it wrote, read from the mutation, `[name, "frame", absolute, relative, size, unit_length]` or `[name, "placement", absolute, relative, direction]`, as TypeScript's suite does, value for value. mbse-basis [CAP-242: incremental physical placement](https://github.com/pitaman71/mbse-basis/blob/main/docs/capabilities/README.md#CAP-242).

```python
from support import Bench
from mbse.Diagrams.Framework import Constraints, Container, Kiwi, Logical, Placement

# A square's four extent points, each a vector of length 2 from its origin: 2 units wide and high.
SQUARE = [("n", 0, 2), ("s", 180, 2), ("e", 90, 2), ("w", 270, 2)]
b = Bench(True, True)
b.tune(unit_length=1, scaling_factor=0.5, margin=0.25)
```

<a id="PLC-01"></a>
## PLC-01 · Each level's unit length: the diagram's, then its container's times the scaling factor

```python
b.child("", "cpu")
b.child("cpu", "alu")
assert [Placement.unit_length(x) for x in (b.overview, b.shape("cpu"), b.shape("cpu.alu"))] == [1.0, 0.5, 0.25]
plain = Bench(True, True)
plain.child("", "x")
assert Placement.unit_length(plain.shape("x")) == 1.0
```

## PLC-11 · A point no vector from its origin reaches lies at its origin, and a vector with no length reaches no farther

```python
odd = Bench(True, True)
odd.child("", "box")
L, s = odd.source.Logical, "s_box"
odd._step("points", ("box", []), ["box"],
          lambda _: getattr(_, s).owns({"origin": lambda e: e.owned(_.origin)}),
          lambda _: _.origin.origin_of([getattr(_, s)]),
          lambda _: getattr(_, s).owns({"lost": lambda e: e.owned(_.lost)}), lambda _: _.lost.border_of([getattr(_, s)]),
          lambda _: getattr(_, s).owns({"short": lambda e: e.owned(_.short)}),
          lambda _: _.short.border_of([getattr(_, s)]),
          lambda _: getattr(_, s).owns({"to short": lambda e: e.owned(_.v)}),
          lambda _: _.v.start([_.origin]).end([_.short]).rotation(lambda r: r.degrees(90.0)),
          spec={"origin": L.OfPoint, "lost": L.OfPoint, "short": L.OfPoint, "v": L.OfVector})
assert odd.placed == '[["origin","placement",[0,0],[0,0],0],["lost","placement",[0,0],[0,0],0],["short","placement",[0,0],[0,0],90]]'
```

## PLC-14 · Following mutation by mutation ends as placing afresh does, through a seeded sequence of every kind of mutation

```python
r = Bench(True, True)
r.tune(unit_length=1.0, scaling_factor=0.5, margin=0.25, minimum_size=1.0)
seed = 2026


def pick(n: int) -> int:
    global seed
    seed = seed * 48271 % 2147483647
    return seed // 65536 % n


kids: dict[str, list[str]] = {"": []}  # each container's children, by handle
kinds: list[str] = []
drift: list[int] = []
for i in range(150):
    boxes = list(kids)
    box = boxes[pick(len(boxes))]
    here = kids[box]
    kind = ["child", "child", "points", "spec", "unspec", "unchild", "turn", "tune", "move"][pick(9)]
    if kind == "child":
        handle = f"{box}.n{i}" if box else f"n{i}"
        r.child(box, f"n{i}")
        here.append(handle)
        kids[handle] = []
    elif kind == "points" and box:
        k = 1 + pick(3)
        r.points(box, [("n", 0, k), ("s", 180, 1 + pick(3)), ("e", 90, k), ("w", 270, 1)])
    elif kind in ("spec", "unspec") and len(here) > 1:
        a, c = here[pick(len(here))], here[pick(len(here))]
        if a == c:
            continue
        if kind == "spec":
            r.spec("", a, c, 90.0 * pick(4))  # by handle
        else:
            r.unspec("", a, c)
    elif kind == "unchild" and box and here:  # to the diagram's top level, keeping its handle
        gone = here.pop(pick(len(here)))
        r.move(gone, "")
        kids[""].append(gone)
    elif kind == "move" and box:  # a top-level shape into a container, never into itself
        top = kids[""]
        if not top:
            continue
        handle = top[pick(len(top))]
        inside, c = False, r.shape(box)
        while not inside and c.as_schema().name != "Logical.OfDiagram":
            inside, c = Logical.name_of(c) == handle, Logical.container_of(c)
        if inside:
            continue
        r.move(handle, box)
        top.remove(handle)
        here.append(handle)
    elif kind == "turn" and box and here:
        r.turn(box, 90.0 * pick(4))
    elif kind == "tune":
        r.tune(margin=0.25 * pick(3))
    else:
        continue
    kinds.append(kind)
    if r.afresh():
        drift.append(i)
assert sorted(set(kinds)) == ['child', 'move', 'points', 'spec', 'tune', 'turn', 'unchild', 'unspec']
assert len(kinds) == 101
assert drift == []
```
